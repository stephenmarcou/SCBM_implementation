```python
import numpy as np
import matplotlib.pyplot as plt

# Example schematic covariance matrix
Sigma = np.array([
    [ 1.0,  0.45, 0.25, 0.30, -0.15, -0.10, -0.20, -0.15],
    [ 0.45, 1.0,  0.50, 0.35, -0.10, -0.20, -0.15,  0.00],
    [ 0.25, 0.50, 1.0,  0.55, -0.05, -0.25, -0.15,  0.05],
    [ 0.30, 0.35, 0.55, 1.0,  -0.20, -0.10, -0.05, -0.10],
    [-0.15,-0.10,-0.05,-0.20, -1.0, -0.45, -0.30, -0.35],
    [-0.10,-0.20,-0.25,-0.10, -0.45,-1.0, -0.50, -0.40],
    [-0.20,-0.15,-0.15,-0.05, -0.30,-0.50, -1.0, -0.55],
    [-0.15, 0.00, 0.05,-0.10, -0.35,-0.40, -0.55, -1.0],
])

C = 4  # number shown as concepts in schematic

fig, ax = plt.subplots(figsize=(3.2, 3.0))

from matplotlib.colors import LinearSegmentedColormap

cmap = LinearSegmentedColormap.from_list(
    "strong_diverging",
    ["#B2182B", "#EF8A62", "#FFFFFF", "#67A9CF", "#2166AC"]
)

im = ax.imshow(
    Sigma,
    cmap=cmap,
    vmin=-1,
    vmax=1,
)

# Divide concept and residual blocks
ax.axvline(C - 0.5, linestyle="--", linewidth=1.3)
ax.axhline(C - 0.5, linestyle="--", linewidth=1.3)

# Remove ordinary ticks
ax.set_xticks([])
ax.set_yticks([])

# c / r block labels
ax.text(C/2 - 0.5, -0.9, r"$\mathbf{c}$",
        ha="center", va="center", fontsize=13)

ax.text(C + (len(Sigma)-C)/2 - 0.5, -0.9, r"$\mathbf{r}$",
        ha="center", va="center", fontsize=13)

ax.text(-0.9, C/2 - 0.5, r"$\mathbf{c}$",
        ha="center", va="center", fontsize=13)

ax.text(-0.9, C + (len(Sigma)-C)/2 - 0.5, r"$\mathbf{r}$",
        ha="center", va="center", fontsize=13)


plt.tight_layout()

plt.savefig(
    "plots_paper/joint_covariance.svg",
    bbox_inches="tight",
    transparent=True
)

plt.show()
```
